```python
import ast
```

```python
def extract_names(text):
    items = ast.literal_eval(text)
    return [item['name'] for item in items]
   
```

```python
movies_data['genres'] = movies_data['genres'].apply(extract_names)
movies_data['keywords'] = movies_data['keywords'].apply(extract_names)
movies_data['cast'] = movies_data['cast'].apply(extract_names)


movies_data[['title', 'genres', 'keywords', 'cast']].head()
```

```python
def extract_director(text):
    items = ast.literal_eval(text)

    for item in items:
        if item['job'] == 'Director':
            return item['name']

    return ''
```

```python
movies_data['director'] = movies_data['crew'].apply(extract_director)

movies_data[['title', 'director']].head()
```

We'll use TF-IDF (Term Frequency–Inverse Document Frequency).

The basic idea is:

*Words that are important for a particular movie get higher values.

*Words that appear in almost every movie get lower values.

*Each movie becomes a numerical vector.

*Later, we'll compare these vectors to find movies with similar content.

We already installed scikit-learn, so we can use its TfidfVectorizer.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words='english'
)

tfidf_matrix = tfidf.fit_transform(model_data['tags'])

tfidf_matrix.shape
```

**Calculate movie similarity**

We'll use cosine similarity for this.

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(tfidf_matrix)

similarity.shape
```

```python
movie_index = model_data[model_data['title'] == 'Avatar'].index[0]

similar_movies = list(enumerate(similarity[movie_index]))

similar_movies = sorted(
    similar_movies,
    key=lambda x: x[1],
    reverse=True
)

similar_movies[:10]
```

```python
for index, score in similar_movies[1:11]:
    print(model_data.iloc[index]['title'], "→", round(score, 3))
```

**create recommend()**

```python
def recommend(movie):
    if movie not in model_data['title'].values:
        print("Movie not found. Please check the movie title.")
        return

    movie_index = model_data[model_data['title'] == movie].index[0]

    similar_movies = list(enumerate(similarity[movie_index]))

    similar_movies = sorted(
        similar_movies,
        key=lambda x: x[1],
        reverse=True
    )

    for index, score in similar_movies[1:6]:
        print(model_data.iloc[index]['title'], "→", round(score, 3))
```

```python
recommend('Avatar')
```

```python
recommend('avatr')
```
